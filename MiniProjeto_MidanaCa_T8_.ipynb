RESUMO DO QUE FOI ENCONTRADO E DO QUE SERÁ FEITO

1. Quantas linhas e colunas existem?

A base possui 830.000 linhas e 14 colunas.

2. O que cada registro representa?

Cada registro contém informações sobre a data da compra,item de compra, o gênero e o estado civil do cliente, além da categoria, do segmento de cliente e do nome do produto. Em alguns registros, a categoria e o nome do produto aparecem identificados como “#N/D”.

3. Quais informações estão ausentes?

O diagnóstico realizado com o método isna() identificou valores ausentes apenas 4 colunas 100% vazias totalizando 3.650 #N/D reconhecidos por essa verificação.

4. Existem informações duplicadas?

Foram identificadas 96.553 linhas integralmente duplicadas.

5. Existem textos preenchidos apenas com espaços?

Não. Todas as colunas de texto analisadas apresentaram zero registros preenchidos apenas com espaços vazios ("").

6. Quais diferenças de escrita foram encontradas?

Nas quatro colunas analisadas — Genero_cliente, Segmento_cliente, Categoria_produto e Nome_produto —, não foram identificadas diferenças de escrita relacionadas às formas de apresentação das letras maiúsculas e minúsculas.

7. Quais decisões ainda precisam ser tomadas?

É necessário decidir quais repetições devem ser removidas, como padronizar os textos e quais registros devem ser considerados nas análises.

Problema escolhido: registros duplicados e colunas vazias.

Problema encontrado: além das repetições e das colunas vazias, existem 3.650 registros sem identificação da categoria e do nome do produto, apresentados como “#N/D”.

Como isso afeta a análise?

As repetições podem aumentar indevidamente a contagem de registros e, caso sejam utilizadas variáveis financeiras, podem inflar as somas dos valores de compra, produzindo resultados superiores aos reais. Da mesma forma, a ausência de informações sobre as categorias e os nomes dos produtos limita a análise e a classificação adequada dos itens.

Decisão de tratamento proposta: A base não tem coluna de quantidade. Linhas idênticas dentro da mesma compra (mesma data, cliente e produto) podem ser o mesmo produto comprado mais de uma vez (duas bananas, por exemplo), e não erro de digitação. Sem um identificador único do item ou uma quantidade, não é possível confirmar que sejam duplicatas indevidas. As instruções pedem eliminar apenas duplicatas confirmadas e não alterar dados válidos. Por isso, a decisão padrão é manter as repetições (cada linha = uma unidade do produto) e eliminar as colunas vazias de 10 a 14.

Também decide-se não padronizar os textos, porque como não apresenta nenhuma diferença da escrita dos textos, por isso resolve-se deixar o texto do jeito que está.

## SPRINT 3 — FUNÇÕES





```python
def resumo_numerico(serie):
  return {
      "contagem_valida": serie.count(),
      "media": serie.mean(),
      "minimo": serie.min(),
      "maximo": serie.max()
  }
```

```python
resumo = resumo_numerico(df["Filho_cliente"])

print("Valores válidos: ", resumo["contagem_valida"])
print("Média: ", round(resumo["media"], 2))
# Acessando com as chaves acentuadas conforme gerado pela função
print("Mínimo: ", resumo.get("minimo", resumo.get("mínimo")))
print("Máximo: ", resumo.get("maximo", resumo.get("máximo")))
```

## SPRINT 4 — ANÁLISE

Filtrar clientes com 4 filhos

```python
filhos_igual_4 = df[df["Filho_cliente"] == 4]

display(filhos_igual_4)

print(
    "Quantidade de registros de clientes com 4 filhos: ",
    len(filhos_igual_4)
)
```

##	Filtros e agrupamentos

Vamos fazer a contagem de registros por categoria e soma dos valores por gênero

```python
# Agrupar a coluna categoria de produtos.

produto_por_categoria = (
    df.groupby("Categoria_produto")
    .size()
    .sort_values(ascending=False)
)
display(produto_por_categoria)
```

ALIMENTOS tem o maior número de registros .

```python
# Agrupar por gênero do cliente
grupos_genero_cliente = df.groupby(
    "Genero_cliente", dropna=False
)["Filho_cliente"]

# Criar resumo por gênero
resumo_por_genero = pd.DataFrame({
    "registros": grupos_genero_cliente.size(),
    "valores_validos": grupos_genero_cliente.count(),
    "soma_valores_conhecidos": grupos_genero_cliente.sum(min_count=1)
})

display(resumo_por_genero)
```

# Traduzir as colunas para facilitar a interpretação

```python
# 1. Criar um dicionário para traduzir os nomes
traducao_colunas = {
    "DATA": "Data",
    "CO_ID": "Compra_id",
    "CL_ID": "Cliente_id",
    "CL_GENERO": "Genero_cliente",
    "CL_EC": "Estado_civil_cliente",
    "CL_FHL": "Filho_cliente",
    "CL_SEG": "Segmento_cliente",
    "PR_ID": "Produto_id",
    "PR_CAT": "Categoria_produto",
    "PR_NOME": "Nome_produto"
}

# 2. Renomear as colunas
df = df.rename(columns=traducao_colunas)

# 3. Exibir os novos nomes
print("\nNomes traduzidos:")
print(df.columns.tolist())

```

# Linhas duplicadas




```python
print(
    "Repetições completas além da primeira ocorrência: ",
    df.duplicated().sum() # marca as repetições da primeira ocorrência
)

print(
    "Linhas que participam de grupos repetidos: ",
    df.duplicated(keep=False).sum() # marca todas as ocorrências dos registros repetidos
)

print(
    "Identificadores de pedidos distintos: ",
    df["Compra_id"].nunique()
)
```

```python
# Visualize as linhas duplicadas
display(
    df.loc[df.duplicated(keep=False)])
```

```python
# verificar os espaços

colunas_texto = [
    "Genero_cliente",
    "Segmento_cliente",
    "Categoria_produto",
    "Nome_produto"

]

for coluna in colunas_texto:
  quantidade_em_branco = (
      df[coluna]
      .astype("string")
      .str.strip()
      .eq("")
      .sum()
  )

  print(coluna, ":", quantidade_em_branco)
```
